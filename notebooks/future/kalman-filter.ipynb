# Gaussian
- Gaussian distribution --> another term for normal distribution
- Gaussian functions --> often used to represent the probability density function of a normally distributed random variable (with expected value $\mu = b$, variance $\sigma^{2} = c^{2}$)  
### Use cases of Gaussian Functions:
- Describe normal distributions
- Signal processing to define Gaussian filters (usually for smoothing time-series data)

### Kalman Filter Notes
- Recursive algorithm used for estimating the state of a dynamic system from noisy measurements. (basically the system is noisy, so we are trying to estimate the denoised state)
- It assumes that the noise in the system and measurements follows a Gaussian distribution (normal).
- Optimal full-state estimator, given Gaussian white noise disturbances (disturbance to system) and measurement noise.

Kalman Filter is designed to be updated over time. It operates in a recursive process, where the current estimate is refined based on new incoming measurements. This means that the filter continuously improves its state estimate as more data becomes available.
